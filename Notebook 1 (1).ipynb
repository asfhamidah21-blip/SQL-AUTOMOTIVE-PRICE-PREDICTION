# ANALISIS HARGA BERDASARKAN MEREK MOBIL DAN TAHUN

```
if '_dntk' in globals():
  _dntk.dataframe_utils.configure_dataframe_formatter('{"sortBy":[],"filters":[],"pageSize":10,"pageIndex":0,"columnOrder":["make","year","harga_rata_rata"],"hiddenColumnIds":[],"columnDisplayNames":[],"conditionalFilters":[],"cellFormattingRules":[],"wrappedTextColumnIds":[]}')
else:
  _deepnote_current_table_attrs = '{"sortBy":[],"filters":[],"pageSize":10,"pageIndex":0,"columnOrder":["make","year","harga_rata_rata"],"hiddenColumnIds":[],"columnDisplayNames":[],"conditionalFilters":[],"cellFormattingRules":[],"wrappedTextColumnIds":[]}'

df_17 = _dntk.execute_sql(
  'SELECT make, year, ROUND(AVG(price)) as harga_rata_rata\nFROM df_12\nWHERE make IN (\'Toyota\',\'BMW\',\'Ford\',\'Honda\')\nGROUP BY make, year\nORDER BY make, year',
  'SQL_DEEPNOTE_DATAFRAME_SQL',
  audit_sql_comment='',
  sql_cache_mode='cache_disabled',
  return_variable_type='dataframe'
)
df_17
```

### Analisis pada tabel diatas, menunjukkan perbandingan harga rata\-rata 4 merek \(BMW, Ford, Honda, Toyota\) berdasarkan tahun produksi 2000\-2025\.

### 1\.  Tren Umum Naik: Keempat merek menunjukkan tren kenaikan harga yang signifikan seiring tahun yang lebih muda\. Contoh BMW tahun 2001 hanya 1\.506, namun di tahun 2025 mencapai 47\.363 atau naik 31x lipat\.

### 2\.  BMW Paling Mahal & Konsisten: BMW konsisten menjadi merek termahal di setiap tahunnya\. Di tahun 2020 BMW sudah 32\.004, sedangkan Ford 18\.378, Honda 15\.142, Toyota \(2014\) 5\.117\. Ini membuktikan merek premium memiliki nilai jual kembali yang lebih tinggi\.

### 3\.  Anomali Harga 1\.500 & Fluktuasi: Pada tahun 2000\-2004 semua merek Ford, Honda, Toyota flat di harga 1\.500, ini memberi tanda batas minimal harga \(floor price\) pada dataset\. Fluktuasi naik turun seperti Ford 2007 2\.371 turun ke 2008 2\.156 dan Honda 2005 1\.798 turun ke 2006 1\.507 disebabkan perbedaan tipe varian dan mil di tahun tersebut, bukan penurunan pasar\.

### Kesimpulannya, tahun produksi adalah faktor utama penentu harga, dengan merek menjadi faktor pembeda kedua\.

### Note: Untuk data harga 1500, ditetapkan minimal harga / harga terendah bertujuan untuk menghindari nilai 0 atau outlier yang mengganggu perhitungan rata\-rata\.
