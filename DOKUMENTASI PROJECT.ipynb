# MENGATASI DATA OUTLIER MENGGUNAKAN METODE IQR

```
if '_dntk' in globals():
  _dntk.dataframe_utils.configure_dataframe_formatter('{}')
else:
  _deepnote_current_table_attrs = '{}'

df_20 = _dntk.execute_sql(
  'WITH quartiles AS (\n    SELECT\n        \'year\' AS kolom,\n        year::DOUBLE AS nilai,\n        NTILE(4) OVER (ORDER BY year) AS kuartil\n    FROM \'Automotive Price Prediction-20260918-042121.csv\'\n\n    UNION ALL\n\n    SELECT\n        \'mileage\' AS kolom,\n        mileage::DOUBLE AS nilai,\n        NTILE(4) OVER (ORDER BY mileage) AS kuartil\n    FROM \'Automotive Price Prediction-20260918-042121.csv\'\n\n    UNION ALL\n\n    SELECT\n        \'engine_hp\' AS kolom,\n        engine_hp::DOUBLE AS nilai,\n        NTILE(4) OVER (ORDER BY engine_hp) AS kuartil\n    FROM \'Automotive Price Prediction-20260918-042121.csv\'\n\n    UNION ALL\n\n    SELECT\n        \'owner_count\' AS kolom,\n        owner_count::DOUBLE AS nilai,\n        NTILE(4) OVER (ORDER BY owner_count) AS kuartil\n    FROM \'Automotive Price Prediction-20260918-042121.csv\'\n\n    UNION ALL\n\n    SELECT\n        \'vehicle_age\' AS kolom,\n        vehicle_age::DOUBLE AS nilai,\n        NTILE(4) OVER (ORDER BY vehicle_age) AS kuartil\n    FROM \'Automotive Price Prediction-20260918-042121.csv\'\n\n    UNION ALL\n\n    SELECT\n        \'mileage_per_year\' AS kolom,\n        mileage_per_year AS nilai,\n        NTILE(4) OVER (ORDER BY mileage_per_year) AS kuartil\n    FROM \'Automotive Price Prediction-20260918-042121.csv\'\n\n    UNION ALL\n\n    SELECT\n        \'brand_popularity\' AS kolom,\n        brand_popularity AS nilai,\n        NTILE(4) OVER (ORDER BY brand_popularity) AS kuartil\n    FROM \'Automotive Price Prediction-20260918-042121.csv\'\n\n    UNION ALL\n\n    SELECT\n        \'price\' AS kolom,\n        price AS nilai,\n        NTILE(4) OVER (ORDER BY price) AS kuartil\n    FROM \'Automotive Price Prediction-20260918-042121.csv\'\n),\n\nstats AS (\n    SELECT\n        kolom,\n        MAX(CASE WHEN kuartil = 1 THEN nilai END) AS q1,\n        MIN(CASE WHEN kuartil = 4 THEN nilai END) AS q3\n    FROM quartiles\n    GROUP BY kolom\n)\n\nSELECT\n    s.kolom,\n    s.q1,\n    s.q3,\n    s.q3 - s.q1 AS iqr,\n    s.q1 - 1.5 * (s.q3 - s.q1) AS batas_bawah,\n    s.q3 + 1.5 * (s.q3 - s.q1) AS batas_atas,\n    SUM(\n        CASE\n            WHEN q.nilai < s.q1 - 1.5 * (s.q3 - s.q1)\n              OR q.nilai > s.q3 + 1.5 * (s.q3 - s.q1)\n            THEN 1\n            ELSE 0\n        END\n    ) AS jumlah_outlier\nFROM stats AS s\nJOIN quartiles AS q\n    ON s.kolom = q.kolom\nGROUP BY\n    s.kolom,\n    s.q1,\n    s.q3\nORDER BY s.kolom;',
  'SQL_DEEPNOTE_DATAFRAME_SQL',
  audit_sql_comment='',
  sql_cache_mode='cache_disabled',
  return_variable_type='dataframe'
)
df_20
```

### Berdasarkan metode IQR, ditemukan outlier pada beberapa variabel numerik seperti brand popularity, engine hp, mileage per year, price, vehicle age, dan year\. Outlier tidak langsung dihapus karena nilai tersebut belum tentu merupakan kesalahan data dan masih dapat merepresentasikan kondisi kendaraan yang sebenarnya\.

Penentuan harga sebaiknya mempertimbangkan kombinasi antara merek dan kondisi kendaraan, bukan hanya menggunakan satu faktor\. Pendekatan ini dapat membantu membuat segmentasi harga yang lebih spesifik\.

## KESIMPULAN:
Analisis menunjukkan bahwa harga kendaraan memiliki hubungan dengan beberapa karakteristik kendaraan, terutama usia, mileage, tenaga mesin, kondisi, riwayat kecelakaan, serta model dan merek kendaraan\. Kendaraan yang lebih muda, memiliki mileage lebih rendah, tenaga mesin lebih tinggi, dan kondisi lebih baik cenderung memiliki harga yang lebih tinggi dalam dataset\. Oleh karena itu, faktor\-faktor tersebut dapat dipertimbangkan dalam proses segmentasi dan penentuan harga kendaraan\.
